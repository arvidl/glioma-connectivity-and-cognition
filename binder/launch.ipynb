# Run the PAT16 notebooks on Binder

Choose **Run → Run All Cells** to display the controls below. Select a notebook (start with **04: cognition**, a small download), then click **Download and verify selected runtime**. After it says **Ready**, open the notebook link and choose **Run All Cells** there to initialize its live controls.

Only the chosen notebook's authentic inputs and saved computations are downloaded. Nothing is fetched when this launcher is opened or when its controls are displayed. Use this launcher again before opening a different notebook. Larger imaging/resampling packs can take much longer on public Binder.

Shut down previous notebook kernels before opening another: public Binder has limited memory. Sessions and edits are temporary; download any work you want to keep. Raw FSL reconstruction, full analyses and paper compilation use the local workflow in the README.

````python
from pathlib import Path
import sys
root = Path.cwd() if (Path.cwd()/"pyproject.toml").exists() else Path.cwd().parent
from pat16.notebook_assets import catalog, ensure_notebook
import ipywidgets as widgets
from IPython.display import display, Markdown
profiles = catalog()["notebooks"]
def size(profile):
    return sum(p["bytes"] for b in profile["bundles"] for p in b["parts"])/2**20
choices = [(f"{key}: {profile['notebook'][3:-6].replace('_', ' ')} — {size(profile):.2f} MiB", key) for key, profile in profiles.items()]
selected = widgets.Dropdown(options=choices, value="04", description="Notebook:", layout=widgets.Layout(width="95%"))
button = widgets.Button(description="Download and verify selected runtime", button_style="primary", layout=widgets.Layout(width="350px"))
output = widgets.Output()
def download(_):
    button.disabled = True
    with output:
        output.clear_output()
        try:
            path = ensure_notebook(selected.value)
            display(Markdown(f"**Ready.** [Open notebook {selected.value}](../notebooks/{path.name}), then run all its cells."))
        except Exception as error:
            print(f"Setup failed: {error}. You can retry; verified existing files are reused.")
            raise
        finally:
            button.disabled = False
button.on_click(download)
display(selected, button, output)
````
